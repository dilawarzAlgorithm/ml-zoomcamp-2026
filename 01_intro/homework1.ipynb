```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv"
df = pd.read_csv(url)
```

```python
# Count unique fuel types in the dataset
fuel_col = [c for c in df.columns if 'fuel' in c.lower() and 'eff' not in c.lower()][0]
print(df[fuel_col].nunique())
```

```python
# Number of columns with missing/null values
print((df.isnull().sum() > 0).sum())
```

```python
# Maximum value in the fuel efficiency column
eff_col = [c for c in df.columns if 'eff' in c.lower()][0]
print(df[eff_col].max())
```

```python
hp_col = [c for c in df.columns if 'hp' in c.lower() or 'horsepower' in c.lower()][0]

# 1. Median before filling
median_before = df[hp_col].median()

# 2. Most frequent value (mode)
mode_val = df[hp_col].mode().iloc[0]

# 3. Median after filling missing values with the mode
median_after = df[hp_col].fillna(mode_val).median()

print(f"Before: {median_before}, After: {median_after}")
# Check whether it increased, decreased, or did not change (No)
```

```python
# 1. Select cars where origin is 'Asia' and pick vehicle_weight and model_year
asia_df = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)

# 2. Get the underlying numpy array
X = asia_df.values

# 3. Compute X^T * X and its inverse
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)

# 4. Target array y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 5. Calculate linear regression weights w = (X^T * X)^(-1) * X^T * y
w = XTX_inv @ X.T @ y

# 6. Sum of elements of w
print("Sum of weights:", w.sum())
```
